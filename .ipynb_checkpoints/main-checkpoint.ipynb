```python
material_values = {
        chess.PAWN: 1,
        chess.KNIGHT: 3,
        chess.BISHOP: 3,
        chess.ROOK: 5,
        chess.QUEEN: 9,
        chess.KING: 0,  # King has no material value but is essential
    }

    # Calculate material score
def material_score(board):
        score = 0
        for piece_type, value in material_values.items():
            score += len(board.pieces(piece_type, chess.WHITE)) * value
            score -= len(board.pieces(piece_type, chess.BLACK)) * value
        return score
```

```python
def calculate_reward(board, previous_material, done):
    """
    Calculate the reward for a move.

    Args:
        board (chess.Board): Current board state.
        previous_material (int): Material value before the move.
        done (bool): True if the game is over, otherwise False.

    Returns:
        float: Reward for the move.
    """
    # Define material values
    material_values = {
        chess.PAWN: 1,
        chess.KNIGHT: 3,
        chess.BISHOP: 3,
        chess.ROOK: 5,
        chess.QUEEN: 9,
        chess.KING: 0,  # King has no material value but is essential
    }

    # Calculate material score
    """def material_score(board):
        score = 0
        for piece_type, value in material_values.items():
            score += len(board.pieces(piece_type, chess.WHITE)) * value
            score -= len(board.pieces(piece_type, chess.BLACK)) * value
        return score"""

    # Check game outcome
    if done:
        if board.is_checkmate():
            return 1.0 if board.turn == chess.BLACK else -1.0  # Reward winning player
        elif board.is_stalemate() or board.is_insufficient_material() or board.is_seventyfive_moves():
            return 0.5  # Draw
        else:
            return 0  # No outcome yet (rare case)

    # Calculate material reward
    current_material = material_score(board)
    material_reward = current_material - previous_material

    # Check reward
    check_reward = 0.05 if board.is_check() else 0

    # Combine rewards
    total_reward = material_reward * 0.1 + check_reward
    return total_reward
```

```python
def train_chess_dqn_with_players():
    dqn = model.create_dqn()
    target_dqn = model.create_dqn()
    target_dqn.set_weights(dqn.get_weights())
    memory = model.ReplayBuffer(model.MEMORY_SIZE)

    epsilon = model.EPSILON_START

    for episode in range(1000):  # Train over 1000 games
        env = chess.Board()
        previous_material = 0
        total_reward = 0

        print(f"Starting Game {episode + 1}")

        while not env.is_game_over():
            # Render the board visually
            svg_board = chess.svg.board(env)
            display(SVG(svg_board))

            # Human Move
            print("Your Move (e.g., e2e4): ")
            human_move = input().strip()
            try:
                human_move = chess.Move.from_uci(human_move)
                if human_move not in env.legal_moves:
                    print("Illegal move. Try again.")
                    continue
            except ValueError:
                print("Invalid move format. Try again.")
                continue

            env.push(human_move)

            if env.is_game_over():
                break

            # AI's Turn
            state = model.encode_board(env)
            legal_moves = list(env.legal_moves)
            action = model.select_action(dqn, state, legal_moves, epsilon)
            env.push(action)

            reward = calculate_reward(env, previous_material, env.is_game_over())
            next_state = model.encode_board(env)
            done = env.is_game_over()

            memory.add((state, action, reward, next_state, done))
            model.train_dqn(dqn, target_dqn, memory, model.BATCH_SIZE)

            previous_material =material_score(env)
            total_reward += reward

        epsilon = max(model.EPSILON_MIN, epsilon * model.EPSILON_DECAY)

        if episode % 10 == 0:
            target_dqn.set_weights(dqn.get_weights())

        print(f"Game {episode + 1} ended with reward {total_reward}")

    return dqn

if __name__ == "__main__":
    trained_model = train_chess_dqn_with_players()
```
