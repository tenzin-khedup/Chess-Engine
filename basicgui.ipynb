```python


def choose_best_move(board):
    best_move = None
    best_score = float('-inf')  # Start with the worst possible score

    for move in board.legal_moves:
        # Simulate the move
        board.push(move)
        score = material_count(board)
        # Revert the move
        board.pop()

        # Update the best move if the score improves
        if score > best_score:
            best_score = score
            best_move = move

    return best_move
```

```python
next_move=choose_best_move(board)
```
